```python
import matplotlib.pyplot as plt
year=[1998,1999,2005,2008,2009]
plt.plot(year,[1,3,5,7,9],'ro--',label="Odd",linewidth=3)
plt.plot(year,[2,4,6,8,0],'g^:',label="Even")
plt.xlabel("No.")
plt.ylabel("Age")
plt.title("Random Quest")
plt.grid(True)
plt.style.use('seaborn-v0_8-bright')
```
